## Recovering and Leveraging Builder Data Using Architectural Waves

### 1. The Value of Builder Identity
A home's builder acts exactly like a brand label in retail. Renowned custom builders or premium national developers command significant price premiums due to perceived material quality, architectural design, and structural warranties. Conversely, builders associated with historic class-action lawsuits can carry a price penalty. Preserving this column—even with 95% missing data—is highly valuable for capturing these extreme valuation outliers.

### 2. Capitalizing on Spatial-Temporal Construction Waves
Housing inventory is rarely built in isolation; it moves in distinct geographic and temporal waves. Production builders typically buy large tracts of land and construct master-planned subdivisions of 50 to 500 homes over a compressed 2-to-3-year window. Because our **Latitude, Longitude, and Year Built** columns are highly populated, we can identify these hidden subdivision clusters. If a house is missing a builder name but sits inside a tight geographic cluster where neighboring homes built in the exact same year list a specific builder, we can safely impute that builder name.

### 3. Decoding the "Missingness" Signal
In data science, missing data is rarely random. For this feature, a missing builder name is a predictive signal in itself:
* **Older Homes (e.g., 1950 and earlier):** A missing builder generally represents a standard historical data loss. 
* **Modern Homes (e.g., 2000 and later):** A missing builder strongly implies a generic, unbranded tract home where the listing agent omitted the name because it does not offer a marketing premium. 

By keeping the missing entries as a distinct category labeled `"Generic/Unknown"` and allowing the model to cross-reference it with the `Year Built` column, tree-based models can automatically map out where the absence of a name implies a lack of a premium.


## Recovering School District Data Using Geography

### The Goal
School districts hold massive predictive power for house evaluation, but our dataset has too many missing values in the school columns. Instead of throwing this valuable information away, we will use our highly populated **Latitude and Longitude** columns to reconstruct the missing data. 

---

### The Process
We will fix the missing data using a two-step geographical approach:

1. **Reverse-Geocoding (Geometric Lookup):** 
   We will take the exact coordinates of each house and map them against official school boundary files (Shapefiles). This tells us exactly which school district the house physically sits in, automatically filling in the blanks.
   
2. **Nearest-Neighbor Imputation (Backup):** 
   For any coordinates that fall just outside boundary lines or fail the lookup, we will use a spatial algorithm to find the closest house with a known school district and assign that same district.

---

### Verification
To ensure our geographic model is accurate, we will cross-reference our results against the rows that *did* have school data originally. If our lookup matches the existing data, we can confidently use the newly filled school feature across all ~500,000 rows.
